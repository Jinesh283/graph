# Finite Difference Error vs Step Size

This notebook plots the truncation error of the **forward**, **backward** and **central** difference formulas for the first derivative.

- Function: f(x) = sin(x)
- Point: x = 0.7
- Exact derivative: f′(x) = cos(x)

| Method | Formula | Order |
|---|---|---|
| Forward | [f(x+h) − f(x)] / h | O(h) |
| Backward | [f(x) − f(x−h)] / h | O(h) |
| Central | [f(x+h) − f(x−h)] / (2h) | O(h²) |

```python
import numpy as np
import matplotlib.pyplot as plt

# Function and its exact derivative
f = np.sin
f_prime = np.cos

# Point and step sizes
x0 = 0.7
h = np.logspace(-4, np.log10(0.6), 200)   # 1e-4 to 0.6

# Finite difference approximations
forward  = (f(x0 + h) - f(x0)) / h
backward = (f(x0) - f(x0 - h)) / h
central  = (f(x0 + h) - f(x0 - h)) / (2 * h)

# Absolute errors
exact = f_prime(x0)
err_forward  = np.abs(forward  - exact)
err_backward = np.abs(backward - exact)
err_central  = np.abs(central  - exact)

# Plot
plt.figure(figsize=(9.5, 5.5), dpi=150)
plt.loglog(h, err_forward,  label="Forward difference error")
plt.loglog(h, err_backward, label="Backward difference error")
plt.loglog(h, err_central,  label="Central difference error")

plt.xlabel("Step size h")
plt.ylabel("Absolute error")
plt.title("First-derivative error vs step size (f(x) = sin x, x = 0.7)")
plt.grid(True, which="both", alpha=0.3)
plt.legend(fontsize=8)
plt.tight_layout()
plt.savefig("finite_difference_error.png")   # saves the image for your slide
plt.show()
```

**Result:** Forward and backward errors fall with slope 1 (O(h)); the central error falls with slope 2 (O(h²)), so it is much more accurate at the same step size h.
